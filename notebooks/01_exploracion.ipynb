```python
import pandas as pd
df = pd.read_csv("trafico_historico.csv", sep=";")
df.head()
```

```python

columnas_muestra = [
    "Año",
    "Mes",
    "Operador",
    "ORIG_1_N",
    "DEST_1_N",
    "NAC",
    "PASAJEROS",
    "Distancia"
]

año_reciente = df["Año"].max()

muestra = df[
    (df["Año"] == año_reciente) &
    (df["PASAJEROS"] > 0)
]

muestra[columnas_muestra].head()
```

```python

df.info()
```

# Pasajeros por año

```python
df.groupby("Año")["PASAJEROS"].sum()
```

# Pasajeros por Operador

```python
df.groupby("Operador")["PASAJEROS"].sum().sort_values(ascending=False).head(10)
```

# Pasajeros por mes

```python
df.groupby("Mes")["PASAJEROS"].sum()
```

# Destinos con mayor cantidad de Pasajeros

```python
top_destinos = (
    df.groupby("DEST_1_N")["PASAJEROS"]
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

top_destinos
```
